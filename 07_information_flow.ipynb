**00_choosing_returns**

```python
# defining base requirement
ADULT_AGE: int = 18  # U.S. age

# Function to check if a person is an adult
# It uses a return type hint indicating a boolean value (True or False)
def is_adult(age: int) -> bool:
    # Check if the given age meets or exceeds the adult age
    if age >= ADULT_AGE:
        return True
    else:
        # if criteria mismatched return false
        return False

def main():
    # user input in str which is converted to integer
    age : str = int(input("How old is this person?: "))
    # print the obtained value by calling function
    print(is_adult(age))

if __name__ == "__main__":
    main()
```

02_in_range

```python
def in_range(n, low, high):
  """
  Returns True if n is between low and high, inclusive.
  high is guaranteed to be greater than low.
  """
  if n >= low and n <= high:
    return True

  return False

def main():
  print(in_range(4,2,6)) # first run is for producing True value
  print(in_range(6,7,8)) # second run is for producing False value

if __name__ == '__main__':
  main()
```

**03_in_stock**

```python
# defining fruits stock in a dictionary
fruits_stock = {"apple":200,"mango":300,"pear":400,"strawberry":500}

# defining function that returns fruit
def get_fruits(fruit):
  # this line of code gets the item out of dictionary and if no item available the default is set to 0 to avoid program crash
  return fruits_stock.get(fruit.lower(),0)

# defining main function to execute program
def main():
  # taking user input for the required fruit
  fruit = input("Enter the fruit name to search: ")
  # calling get_fruits function which takes fruit as parameter
  stock = get_fruits(fruit)
  # check if the stock is 0 then print the message to user for unavailablity
  if stock <= 0:
    print("This fruit is not in stock.")
  # otherwise show user the availablilty of the fruit with quantity
  else:
    print(f'This fruit is in stock! Here is how many: \n{stock}')

if __name__ == '__main__':
  main()
```

**04_multiple_returns**

```python
def get_user_data():
  first_name = input("Enter your first name: ")
  last_name = input("Enter your last name: ")
  email = input("Enter your email address: ")
  # Returning multiple values separated by commas creates a tuple
  return first_name,last_name,email

def main():
  user_data = get_user_data()
  print(f'Received following user data:\n{user_data}')

if __name__ == '__main__':
  main()
```
